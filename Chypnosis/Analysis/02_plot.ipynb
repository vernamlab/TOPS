# Plot Results

This notebook is the cleaned post-processing entry point for the ChipWhisperer frequency/voltage sweep data.

## Expected input
- Export the `points` list from `01_sweep.ipynb`.
- Save that export as `freq_volt3.txt` in this folder.
- The file is not checked into the repo because it is generated experiment data.


## How to run
1. Run the CW sweep notebook and save the resulting `points` list.
2. Place the export at `Plot_results/freq_volt3.txt`.
3. Run the cells below to preview the data and generate plots.

The same sweep export can be used both for clock-health visualization and for impedance-oriented analysis.


```python
from pathlib import Path
import sys

WORK_DIR = Path.cwd()
if (WORK_DIR / "plot_results.py").exists():
    sys.path.insert(0, str(WORK_DIR))
elif (WORK_DIR / "Plot_results" / "plot_results.py").exists():
    sys.path.insert(0, str((WORK_DIR / "Plot_results").resolve()))

from plot_results import build_heatmap, load_points, metric_dataframe

DATA_FILE = WORK_DIR / "freq_volt3.txt"
if not DATA_FILE.exists():
    DATA_FILE = WORK_DIR / "Plot_results" / "freq_volt3.txt"
if not DATA_FILE.exists():
    raise FileNotFoundError(
        "Save the exported CW sweep data as Plot_results/freq_volt3.txt before running this notebook."
    )
```

```python
points = load_points(DATA_FILE)
clock_frame = metric_dataframe(points, "clock_count")
recovery_frame = metric_dataframe(points, "clock_counter_rec")
register_frame = metric_dataframe(points, "reg_assign_rec")

print(f"Loaded {len(points)} sweep points from {DATA_FILE}.")
clock_frame.head()
```

```python
clock_heatmap = build_heatmap(clock_frame)
recovery_heatmap = build_heatmap(recovery_frame)
register_heatmap = build_heatmap(register_frame)

clock_heatmap.iloc[:5, :5]
```
